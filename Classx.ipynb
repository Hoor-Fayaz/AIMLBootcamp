```python
titanic_dropped = titanic.drop(columns = ["Cabin"])
titanic_dropped.isnull().sum()
```

```python
# mean

# median

titanic_dropped["Age"].fillna(
    titanic_dropped["Age"].mean(), inplace = True
)
```

```python
titanic_dropped.isnull().sum()
```

```python
# mode


titanic_dropped["Embarked"].fillna(
    titanic_dropped["Embarked"].mode()[0], inplace = True
)
```
