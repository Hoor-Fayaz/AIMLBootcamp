```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "/kaggle/input/datasets/hoorawan/iiiiiirrris/iris.csv"
)
```

```python
X = df.iloc[:,0:4].values
y = df["species"].values
```

```python
means = np.mean(X,axis = 0)
std = np.std(X, axis = 0)

print(means)
print(std)
```

```python
max = np.max(X, axis=0)
min = np.min(X, axis =0)

print(max)
print(min)
```

```python
long = X[X[:,2]>5]
print(long)
```

```python
for species in np.unique(y):
    group = X[y ==species]
    print(species)
    print("mean  petal length" , group[:,2].mean())
    print("std of petal length",  group[:,2].std())
    
```
